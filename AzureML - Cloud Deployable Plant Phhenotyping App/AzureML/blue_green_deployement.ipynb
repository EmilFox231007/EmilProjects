```python
from azure.ai.ml.entities import KubernetesOnlineDeployment, CodeConfiguration, ResourceSettings, ResourceRequirementsSettings

model_name = "axonrooter-finetuned"
env = ml_client.environments.get(name="my-custom-env-2", version="1.11")

code_path = r"C:\Users\emilp\Documents\GitHub\2024-25d-fai2-adsai-group-cv9\AxonRooter\src\axonrooter"

blue_deployment = KubernetesOnlineDeployment(
    name="blue",
    endpoint_name=online_endpoint_name,
    model="axonrooter-finetuned:8",  # Static known version
    environment=env,
    code_configuration=CodeConfiguration(
        code=code_path,
        scoring_script="scoring.py"
    ),
    instance_count=1,
    resources=ResourceRequirementsSettings(
        requests=ResourceSettings(cpu="8", memory="15Gi")
    )
)

ml_client.begin_create_or_update(blue_deployment).result()
```

```python
# You can execute this after the blue deployment succeeds

latest_model = max(
    ml_client.models.list(name=model_name),
    key=lambda m: int(m.version)
)

green_deployment = KubernetesOnlineDeployment(
    name="green",
    endpoint_name=online_endpoint_name,
    model=latest_model,  # Automatically gets the newest version
    environment=env,
    code_configuration=CodeConfiguration(
        code=code_path,
        scoring_script="scoring.py"
    ),
    instance_count=1,
    resources=ResourceRequirementsSettings(
        requests=ResourceSettings(cpu="100m", memory="0.5Gi")
    )
)

ml_client.online_deployments.begin_create_or_update(green_deployment).wait()
print("✅ Green deployment created.")
```
