```python
from axonrooter import run_inference, load_and_preprocess_image, preprocess_closing, apply_strict_bboxes, extract_bottom_tips
import matplotlib.pyplot as plt
```

```python
padded_image, pad_top, pad_bottom, pad_left, pad_right = load_and_preprocess_image(f"C:/Users/Endijs/Desktop/School/Y2/Block D/Testing images/test1.png")

# === Show the Preprocessed Image ===
plt.imshow(padded_image, cmap='gray')
plt.title("Padded Preprocessed Image")
plt.axis('off')
plt.show()
```

```python
predicted_mask = run_inference(padded_image, pad_top, pad_bottom, pad_left, pad_right)

# === Show the Preprocessed Image ===
plt.imshow(predicted_mask, cmap='gray')
plt.title("Predicted Mask Image")
plt.axis('off')
plt.show()
```

```python
closed_mask = preprocess_closing(predicted_mask)
```

```python
final_labeled_mask, filtered_bboxes = apply_strict_bboxes(
    closed_mask,
    predicted_mask,
    iou_threshold=0.001,
    size_threshold=2000,
    bottom_y_cutoff_ratio=0.85,
    top_y_cutoff=100,
    min_x_distance=20
)
```

```python
tips = extract_bottom_tips(final_labeled_mask)
```
